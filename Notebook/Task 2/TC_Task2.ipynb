## Business Questions 
This section contains 10 business questions that are answered using SQL

## Business Question 1

How many customers are present in the dataset?

```python
query = """
SELECT COUNT(*) AS total_customers
FROM customers
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 2

How many customers have churned?

```python
query = """
SELECT COUNT(*) AS churned_customers
FROM customers
WHERE Churn = 'Yes'
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 3

What is the average monthly charge?

```python
query = """
SELECT AVG(MonthlyCharges) AS average_monthly_charge
FROM customers
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 4

How many customers are there in each contract type?

```python
query = """
SELECT Contract,
       COUNT(*) AS customer_count
FROM customers
GROUP BY Contract
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 5

Which contract type has the highest number of churned customers?

```python
query = """
SELECT Contract,
       COUNT(*) AS churned_customers
FROM customers
WHERE Churn = 'Yes'
GROUP BY Contract
ORDER BY churned_customers DESC
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 6

What is the average tenure of customers?

```python
query = """
SELECT AVG(tenure) AS average_tenure
FROM customers
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 7

Which internet service is used by the most customers?

```python
query = """
SELECT InternetService,
       COUNT(*) AS customer_count
FROM customers
GROUP BY InternetService
ORDER BY customer_count DESC
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 8

Which payment method is most commonly used?

```python
query = """
SELECT PaymentMethod,
       COUNT(*) AS customer_count
FROM customers
GROUP BY PaymentMethod
ORDER BY customer_count DESC
"""
result = pd.read_sql(query, engine)
result
```

## Business Question 9

Which customers pay more than the average monthly charge?

```python
query = """
SELECT customerID,
       MonthlyCharges
FROM customers
WHERE MonthlyCharges >
(
    SELECT AVG(MonthlyCharges)
    FROM customers
)
ORDER BY MonthlyCharges DESC
"""
result = pd.read_sql(query, engine)
result.head(10)
```

## Business Question 10

What is the ranking of customers based on monthly charges?

```python
query = """
SELECT
    customerID,
    MonthlyCharges,
    RANK() OVER (
        ORDER BY MonthlyCharges DESC
    ) AS charge_rank
FROM customers
"""
result = pd.read_sql(query, engine)
result.head(10)
```
